```python
import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set(style="darkgrid")
%matplotlib inline
```

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
regions = gpd.read_file('/content/drive/MyDrive/Capstone - data/TSDM/District_Boundary.shp')
regions.plot(figsize=(40,20))
regions.sample(5)
```

```python
regions['Dist_Name']
```

```python
data = pd.read_excel('/content/drive/MyDrive/Capstone - data/Temp 2020/Temp 2020 present.xlsx')
data.head(35)
```

```python
merged = regions.set_index('Dist_Name').join(data.set_index('District'))
merged.head(5)
```

```python
fig, ax = plt.subplots(1, figsize=(40, 20))
ax.axis('off')
ax.set_title('Heat Map of Telangana temp(⁰C) across various districts', fontdict={'fontsize': '40', 'fontweight' : '3'})
color = 'Reds'
vmin, vmax = 20,40
sm = plt.cm.ScalarMappable(cmap=color, norm=plt.Normalize(vmin=vmin, vmax=vmax))
sm._A = []
cbar = fig.colorbar(sm)
cbar.ax.tick_params(labelsize=20)
merged.plot('Temp Max (°C)', cmap=color, linewidth=0.8, ax=ax, figsize=(40,20),edgecolor=".4")
```

```python
fig, ax = plt.subplots(1, figsize=(40, 20))
ax.axis('off')
ax.set_title('Heat Map of Rainfall in Telangana across various districts', fontdict={'fontsize': '40', 'fontweight' : '3'})
color = 'Reds'
vmin, vmax = 1,6
sm = plt.cm.ScalarMappable(cmap=color, norm=plt.Normalize(vmin=vmin, vmax=vmax))
sm._A = []
cbar = fig.colorbar(sm)
cbar.ax.tick_params(labelsize=20)
merged.plot('Rain (mm)', cmap=color, linewidth=0.8, ax=ax, figsize=(40,20),edgecolor=".4")
```
