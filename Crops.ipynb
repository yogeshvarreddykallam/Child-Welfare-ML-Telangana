```python
import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set(style="darkgrid")
%matplotlib inline
```

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
regions = gpd.read_file('/content/drive/MyDrive/Capstone - data/TSDM/District_Boundary.shp')
regions.plot(figsize=(40,20))
regions.sample(5)
```

```python
data = pd.read_excel('/content/drive/MyDrive/Capstone - data/Area under different Crops (2).xlsx')
data.head(40)
data.fillna(0)
```

```python
regions['Dist_Name'] = regions['Dist_Name'].str.upper()
```

```python
merged = regions.set_index('Dist_Name').join(data.set_index('Districts'))
merged.tail(40)
```

```python
fig, ax = plt.subplots(1, figsize=(40, 20))
ax.axis('off')
ax.set_title('Heat Map of Area of Rice(In Hectares)', fontdict={'fontsize': '40', 'fontweight' : '3'})
color = 'YlGn'
vmin, vmax = 0,220000
sm = plt.cm.ScalarMappable(cmap=color, norm=plt.Normalize(vmin=vmin, vmax=vmax))
sm._A = []
cbar = fig.colorbar(sm)
cbar.ax.tick_params(labelsize=20)
merged.plot('Rice', cmap=color, linewidth=0.8, ax=ax, figsize=(40,20),edgecolor=".4")
```
